## Ejercicio 1: Operaciones fundamentales con NumPy

### Arreglo 1D: forma y tipo de dato

```python
arreglo_1d = np.array([5, 10, 15, 20, 25, 30, 35, 40, 45, 50, 55, 60])
print(arreglo_1d)
print("Forma:", arreglo_1d.shape)
print("Tipo de dato:", arreglo_1d.dtype)
```

### Cambio de forma con reshape()

```python
matriz_2d = arreglo_1d.reshape(4, 3)
print(matriz_2d)
print("Nueva forma:", matriz_2d.shape)
```

### Matrices aleatorias con np.random.rand() y np.random.randint()

```python
np.random.seed(7)
matriz_rand = np.random.rand(2, 4)
print(matriz_rand)
```

```python
matriz_randint = np.random.randint(10, 50, size=(3, 4))
print(matriz_randint)
```

### Matrices estructurales: np.zeros(), np.ones() y np.eye()

```python
ceros = np.zeros((2, 4))
unos = np.ones((3, 2))
identidad = np.eye(4)
print("Ceros:")
print(ceros)
print("Unos:")
print(unos)
print("Identidad:")
print(identidad)
```

### Transpuesta con .T

```python
matriz_aleatoria = np.random.randint(1, 20, size=(3, 2))
print(matriz_aleatoria)
print("Forma original:", matriz_aleatoria.shape)
print(matriz_aleatoria.T)
print("Forma transpuesta:", matriz_aleatoria.T.shape)
```

### Concatenación: np.hstack() y np.vstack()

```python
a = np.array([[9, 8], [7, 6]])
b = np.array([[1, 0], [3, 2]])
print("Horizontal:")
print(np.hstack((a, b)))
print("Vertical:")
print(np.vstack((a, b)))
```

### Operaciones matemáticas

```python
datos = np.array([15, 4, 22, 9, 31, 18, 6, 27])
print("Suma:", np.sum(datos))
print("Promedio:", np.mean(datos))
print("Mediana:", np.median(datos))
print("Producto:", np.prod(datos))
print("Desviación estándar:", np.std(datos))
print("Varianza:", np.var(datos))
print("Mínimo:", np.min(datos))
print("Máximo:", np.max(datos))
print("Suma acumulativa:", np.cumsum(datos))
```

### Operaciones elemento a elemento

```python
x = np.array([5, 10, 15, 20])
y = np.array([2, 4, 6, 8])
print("Suma:", x + y)
print("Resta:", x - y)
print("Multiplicación:", x * y)
```

## Ejercicio 2: Integración y manipulación con Pandas

```python
import pandas as pd
import numpy as np
```

### DataFrame a partir de un arreglo 2D de NumPy

```python
arreglo = np.array([[2, 8, 14], [5, 11, 17], [9, 3, 20], [7, 13, 4], [16, 10, 1]])
df_array = pd.DataFrame(arreglo, columns=["A", "B", "C"])
df_array
```

### DataFrame a partir de un diccionario

```python
diccionario = {
    "ID": ["0001", "0002", "0003", "0004", "0005", "0006"],
    "Nombre": ["Ian", "Fabian", "Maicol", "Samuel", "Fabian", "Kevin"],
    "Edad": [20, 22, 19, 23, 21, 24],
    "Ciudad": ["Anserma", "Manizales", "Pereira", "Armenia", "Viterbo", "Medellín"]
}
df_dict = pd.DataFrame(diccionario)
df_dict
```

### De DataFrame a arreglo de NumPy: .to_numpy() y .values

```python
arreglo_numpy = df_array.to_numpy()
print(arreglo_numpy)
print(type(arreglo_numpy))
```

```python
arreglo_values = df_array.values
print(arreglo_values)
print(type(arreglo_values))
```

### Promedios con np.mean() sobre el DataFrame

```python
print("Promedio de cada columna:")
print(np.mean(df_array, axis=0))
print("Promedio de cada fila:")
print(np.mean(df_array, axis=1))
```

### Promedios con los métodos nativos del DataFrame

```python
print("Promedio de cada columna:")
print(df_array.mean(axis=0))
print("Promedio de cada fila:")
print(df_array.mean(axis=1))
```
